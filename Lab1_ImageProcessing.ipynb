## Task 2: Loading and Visualizing Images

### A. Loading and displaying an image using OpenCV

```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread('images/iphone_16_wallpaper1.jpg')

plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.show()
```

### B. Loading and displaying an image using PIL

```python
from PIL import Image
import os
import matplotlib.pyplot as plt

files = os.listdir('images')

second_image = [f for f in files if f != 'iphone_16_wallpaper1.jpg'][0]

img2 = Image.open('images/' + second_image)

plt.imshow(img2)
plt.show()
```
